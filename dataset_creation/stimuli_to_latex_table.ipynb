# Convert Stimuli List to Latex Table
Read the stimuli from the csv file and convert them all into a Latex table

```python
import pandas as pd
```

```python
df = pd.read_csv("dataset_experiment_stimuli.csv", sep=";")

# Select columns
df = df[
    [
        # Experiment;Item_type;Stimulus;Condition;Length;Lg10SUBTLEX_US;correct_response

        "Experiment",
        "Item_type",
        "Stimulus",
        "Condition",
        "Length",
        "Lg10SUBTLEX_US",
        "correct_response"
    ]
]

# Convert each row to LaTeX
rows = []

for _, row in df.iterrows():
    values = [
        str(row["Experiment"]),
        str(row["Item_type"]),
        str(row["Stimulus"]),
        str(row["Condition"]),
        str(row["Length"]),
        f'{row["Lg10SUBTLEX_US"]:.2f}',
        str(row["correct_response"])
    ]

    rows.append(" & ".join(values) + r" \\")

print("\n".join(rows))
```
