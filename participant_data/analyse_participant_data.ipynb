# Analyse Participant Data

This code analyses particpant data (means, standard deviations, ranges) and returns the info formatted for a Latex table.

```python
import pandas as pd
```

```python
df = pd.read_csv("questionnaire.csv")

education_map = {
	"No school degree": 1,
	"Lower secondary school": 2,
	"Intermediate secondary school (Mittlere Reife)": 3,
	"Higher education entrance qualification (Abitur / A-level equivalent)": 4,
	"Bachelor’s degree": 5,
	"Master’s degree / Diploma": 6,
	"Doctoral degree (PhD)": 7,
	
}

variables = {
	"age": "Age (years)",
	"age_of_acquisition": "Age of English acquisition (years)",
	"education": "Educational level",
	"proficiency_speaking": "Proficiency Speaking (self-assessed)",
	"proficiency_listening": "Proficiency Listening (self-assessed)",
	"proficiency_reading": "Proficiency Reading (self-assessed)",
	"proficiency_writing": "Proficiency Writing (self-assessed)",
	"usage_speaking": "Usage Speaking",
	"usage_listening": "Usage Listening",
	"usage_reading": "Usage Reading",
	"anxiety_score": "Anxiety score",
	"openness_to_experience": "Openness to experience",
	"conscientiousness": "Conscientiousness",
	"extraversion": "Extraversion",
	"agreeableness": "Agreeableness",
	"neuroticism": "Neuroticism",
	
}
for column, label in variables.items():
	# special case: education level
	if column == "education":
		x = df[column].map(education_map).dropna()
		# print (x)

	# all others
	else:
	# x = pd.to_numeric(df[column], errors="coerce").dropna()			# doesnt work bc of commas
		x = pd.to_numeric(
			df[column].astype(str).str.replace(",", ".", regex=False),
			errors="coerce"
		).dropna()

	n = x.count()

	mean = x.mean()
	median = x.median()
	sd = x.std()
	minimum = x.min()
	maximum = x.max()
	

	

	print(
		f"{label} & "
		f"{mean:.2f} & "
		# f"{median:.2f} & "
		f"{sd:.2f} & "
		f"{minimum:g} -- {maximum:g} \\\\ % {column}"
	)
print(f"\n(Statistics over {n} participants)")
```
